## 2.6: Calculating the root mean squared error

One use of the residuals is to use them to tell us how well the regression line fits the data. We can do this by computing the "mean squared residual" (MSE) using the formula: 

$$MSE = \frac{1}{n}\sum_{i=1}^n {(y_i - \hat{y_i})^2}$$

If we take the square root of the MSE, we get the "root mean squared residual" (RMSE) which tells us how far, on average, the predictions from the regression line are from the actual y values. The RMSE is in the same units as the original y variable, so in our case it will be in runs.

We will still use the common abbreviation RMSE, but we will call this the "root mean squared residual" instead of the "root mean squared error".

## 2.8: Relating the RMSE to the correlation coefficient r

In the first part of this lab we showed that the correlation coefficient, r, can be used to assess how useful particular baseball statistics are in terms of characterizing the batting abilities of a hitter and in the previous exercise you used root mean squared residuals as a measure to determine how good a statistic is at predicting runs. When comparing these two methods, you should have noticed that the order of the statistics based on correlation coefficient and the order of the statistics based on the RMSE were the same. This was not coincidence! It turns out that using these two methods are equivalent since the correlation coefficient (r) and the MSE are related by the formula: 

$$r^2 =  1 - \frac{MSE}{s_y}$$

where: 

- $r^2$: is the squared correlation coefficient
- MSE: is the mean squared error
- $s_y$: is the variance of y computed by a method that does not correct for bias, i.e. $s_y = \frac{1}{n}\sum_i^n (y_i - \bar{y}^2)$. This variance formula that does not correct for bias is the formula used by the numpy's `np.var()` function. 

Note: almost all statisticians use the bias corrected formula for the variance which is: $s_y = \frac{1}{n-1}\sum_i^n (y_i - \bar{y}^2)$, but for some reason numpy has decided to use the biased formula. For reasonably large n, however, the difference is negligible.

# Part 3: Multiple regression

In multiple linear regression we use multiple predictions $x_{ij}$ and multiple coefficients $\hat{\beta_j}$ to predict a value $y_{i}$, i.e., we use data to find the $\hat{\beta_j}$ in the equation $\hat{y_i} = \hat{\beta_0} + \hat{\beta_1} \cdot x_{i1} + \hat{\beta_2} \cdot x_{i2} + ... + \hat{\beta_k} \cdot x_{ik}$. Once we have "learned" the coefficients $\hat{\beta_j}$ from a dataset (usually by minimizing the RMSE) we can then use this equation to make new predictions $\hat{y_i}$ for different $x_i$.

In the following exercise, you will use multiple regression to try to learn a new statistic that can predict runs better than all the other statistics we have seen so far (i.e., better than OPS).
