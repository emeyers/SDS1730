# Part 2. Array and Series operations

Two fundamental packages in the Python data science ecosystem are [numpy](https://numpy.org/) and [pandas](https://pandas.pydata.org/). Numpy is a package that provides support for working with arrays, while pandas is a package that provides support for working with tabular data (i.e., data in tables). In this part of the lab, we will practice some basic calculations using numpy and pandas. 

Numpy arrays are similar to lists, but all elements in a numpy array must be of the same type (e.g., all integers, or all floats). Numpy arrays are more efficient than lists for numerical calculations because they are implemented in C and use contiguous memory locations. This means that numpy arrays can be processed more quickly than lists, especially when working with large datasets. 

Pandas Series are similar to numpy arrays, but they have an index that allows you to label the elements in the Series. This makes it easier to work with data that has a specific meaning, such as time series data or data with categorical labels. Pandas Series also have many built-in methods for performing calculations and transformations on the data. 

In the following exercise we will practice using numpy arrays and pandas Series to perform calculations on baseball data. To calculate the number of singles, we can subtract the number of doubles, triples, and home runs from the total number of hits. 

We can extract Series from a DataFrame by using the syntax `my_df["col_name"]`, where `my_df` is the DataFrame and `col_name` is the name of the column we want to extract. Also, if we have two Series, we can perform element-wise operations on them using the `+`, `-`, `*`, and `/` operators. For example, if we have two Series called `s1` and `s2`, we can add them together using `s1 + s2`.

## 2.2 Get season level slugging percentage

We can also use pandas Series to calculate derived statistics such as batting average (AVG) and slugging percentage (SLG). Batting average is calculated as the number of hits divided by the number of at-bats, while slugging percentage is calculated using the formula: $$SLG = \frac{1B + 2 \times 2B  + 3 \times 3B  + 4 \times HR }{AB}$$ where $1B$ is the number of singles, $2B$ is the number of doubles, $3B$ is the number of triples, and $HR$ is the number of home runs.
