## 1. Writing functions

Functions are useful when writing code one wants to reuse. We can create functions in Python using a "def" statement. Let's try it now!


### 1.1 Writing a function to the ERA

Last class we used numpy arrays to calculate a pitcher's ERA from the number of innings pitched and the number of earned runs they gave up using the formula: $$ERA = \frac{earned\_runs}{innings\_pitched} \times 9$$. Let's now turn this code into a function that calculates ERA for us!

The function we will write is called `get_ERA(innings_pitched, earned_runs)` and takes the following arguments: 

1. `innings_pitched`: A sequence showing how many inning a pitcher pitched each year
2. `earned_runs`: A sequence showning how many earned runs a pitcher gave up each year

The function should return two values (as a tuple): 
1. The ERA for each season as a numpy array
2. The total ERA based on all the data as a number



```python
import numpy as np


# Answer

def calc_ERA(innings_pitched, earned_runs):

    # calculate ERA for each season  
    
    earned_runs_array = np.array(earned_runs)  # convert to arrays
    innings_pitched_array = np.array(innings_pitched)
    
    season_ERA = (earned_runs_array / innings_pitched_array) * 9  # calculate ERA
    season_ERA = np.round(season_ERA, 2)

    
    # calculate the total ERA
    total_ERA = np.sum(earned_runs_array)/np.sum(innings_pitched_array) * 9

    return (season_ERA, float(total_ERA))




# Try out the function

# Lists showing Ohtani's seasons, earned runs, and innings pitched
seasons = [2018, 2020, 2021, 2022, 2023, 2025]
earned_runs = [19, 7, 46, 43, 46, 15]
innings_pitched = [51.667, 1.667, 130.333, 166.0, 132.0, 47.0]


# Use tuple unpacking to get ERA for each season and overall ERA
Ohtani_season_ERA, Ohtani_overall_ERA = calc_ERA(innings_pitched, earned_runs)


print(Ohtani_season_ERA)
print(Ohtani_overall_ERA)

```

### 2.10 Joining/merging DataFrames

We can combine DataFrames together using the `.join()` and `.merge()` methods. The `join()` method joins to DataFrames based on the Index values, while the `.merge()` method joins two DataFrames based on values in columns in the two DataFrames. If we have two DataFrames called `df1` and `df2` that have the same Index values, we can join the using `df1.join(df2)`. Likewise, if the two DataFrame have columns with similar values (but difference Indexes) we can join them using `df1.merge(df2, left_on = "col_df1", right_on = "col_df2)`. We can also use the `how` argument to do, left, right, inner and full joins.
