# Part 1: Correlation

Correlation is a statistical measure that describes the strength and direction of the linear relationship between two variables. The correlation coefficient, denoted as $r$, ranges from -1 to 1. A correlation of -1 indicates a perfect negative linear relationship, a correlation of 0 indicates no linear relationship, and a correlation of 1 indicates a perfect positive linear relationship. 

Let's visualize data using scatter plots and calculate the correlation between home runs and wins to explore the relationship between these two variables.


## 1.1 Scatter plot to visualize the relationship between home runs and wins

Let's use the `plt.scatter()` function to create a scatter plot of the relationship between runs (R) and wins (W). We will set the `alpha` parameter to 0.25 to make the points more transparent so that we can see the density of the points better. We will also add labels to the x and y axes.


```python

plt.scatter(teams_162.R, teams_162.W, alpha = .25);
plt.xlabel("Runs");
plt.ylabel("Wins");
```

## 1.2 Correlation

Let's calculate the correlation between runs and wins using the Series `corr()` method. This will give us a value between -1 and 1 that indicates the strength and direction of the linear relationship between the two variables. 

Please guess what you believe the correlation is from looking at the scatter plot above and then calculate the correlation to see if your guess was close. 

You can also use the `np.corrcoef()` function to calculate the correlation between the two variables. This function will return a 2x2 matrix where the value in the first row and first column is the correlation between the first variable and itself (which is always 1), the value in the second row and second column is the correlation between the second variable and itself (which is also always 1), and the value in the first row and second column (or second row and first column) is the correlation between the two variables.


```python

# calculate the correlation between home runs and wins
print(teams_162.R.corr(teams_162.W))

# using the np.corrcoef() function to calculate the correlation between home runs and wins
print(np.corrcoef(teams_162.R, teams_162.W))
```

# Part 2: Simple linear regression

Regression is a statistical method for modeling the relationship between a dependent variable and one or more independent variables. In simple linear regression, we model the relationship between a dependent variable and a single independent variable using a linear equation. 

The equation for a simple linear regression model is: $\hat{y} = b_0 + b_1 x$

where 

- $\hat{y}$ is the predicted value of the dependent variable
- $b_0$ is the intercept
- $b_1$ is the slope
- $x$ is the independent variable.

Let's explore this now!


## 2.1 Regression model

Let's use the `statsmodels` package to fit a simple linear regression model where the dependent variable is wins (W) and the independent variable is runs (R). We will use the `ols` function from the `statsmodels.formula.api` module to fit the model. The formula for the model will be `"W ~ R"`, which means that we are modeling wins as a function of runs. 

After fitting the model, let's print out what the coefficients are for the intercept and the slope. Can you write the regression equation in the form of $\hat{y}_W = b_0 + b_1*R$, where $b_0$ is the intercept and $b_1$ is the slope? 

Also, please answer these questions:

1. Approximately how many additional runs do you need to score for an additional win? 
2. How many wins will you have if you score 0 runs all season? 


```python

import statsmodels.formula.api as smf

lm = smf.ols(formula = "W ~ R", data = teams_162).fit()

# print the summary of the model

coefficients = lm.params

print(coefficients)

print(1/coefficients["R"])
```

**Answers**

The regression equation is $\hat{y}_w =  25 + 0.076 \cdot x_R$,  
where $y_w$ is wins and $x_R$ is runs. 

1. For the first question, we can calculate the number of additional runs needed for an additional win by taking the reciprocal of the slope coefficient. The slope coefficient is 0.076, so the number of additional runs needed for an additional win is approximately 13.16 runs (1/0.076). 

2. For the second question, we can calculate the number of wins if you score 0 runs which is ~25 wins (although this is not realistic since it is outside of the range that the data was fit)




## 2.2 Making predictions 

We can use the fitted regression model to make predictions about the number of wins for a team based on the number of runs they score. To do this, we can create a new DataFrame with the values of runs that we want to predict wins for, and then we can use the `predict` method of the fitted model to get the predicted wins for those values of runs.


```python

predicted_points = pd.DataFrame({
    "R": [600, 700, 800]
})


predicted_wins = lm.predict(predicted_points)

predicted_wins

```

## 2.3 Residuals 

Residuals are the difference between the observed values of the dependent variable (in this case, wins) and the predicted values from the regression model. We can write this as: 

$e_i = y_i - \hat{y}_i$

where:
- $e_i$ is the residual for the $i$-th observation
- $y_i$ is the observed value of wins for the $i$-th observation
- $\hat{y}_i$ is the predicted value of wins from the regression model for the $i$-th observation. 


Let's calculate the residuals for the first 5 teams in the `teams_162` DataFrame. We will first use the `predict` method of the fitted model to get the predicted wins for these teams, and then we will calculate the residuals by subtracting the predicted wins from the actual wins.

What do these values tell us? 


```python

first_5_teams = teams_162.head(5)

display(first_5_teams[["yearID", "teamID", "R", "W"]])

first_5_team_predicted_wins = lm.predict(first_5_teams)

first_5_team_residuals = first_5_teams.W - first_5_team_predicted_wins

first_5_team_residuals
```

## 2.4 Root mean squared error (RMSE)

The root mean squared error (RMSE) is a measure of the average magnitude of the residuals. It is calculated as the square root of the average of the squared residuals. The formula for RMSE is: $RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} e_i^2}$, where $e_i$ is the residual for the $i$-th observation and $n$ is the number of observations. 

Let's calculate the RMSE for the first 5 teams using the residuals we calculated in the previous step. 


```python

print(np.sqrt(np.mean(first_5_team_residuals ** 2)))

```

# Part 3: Multiple regression 

In multiple regression, we model the relationship between a dependent variable and multiple independent variables using a linear equation. The equation for a multiple regression model is: $\hat{y} = b_0 + b_1 x_1 + b_2 x_2 + ... + b_k x_k$, where $\hat{y}$ is the predicted value of the dependent variable, $b_0$ is the intercept, $b_1$ is the slope for the first independent variable, $b_2$ is the slope for the second independent variable, and so on. 



## 3.1 Fitting a multiple regression model

Let's fit a multiple regression model where the dependent variable is wins (W) and the independent variables are runs scored (R) and runs allowed.  To do this, we will use the `ols` function from the `statsmodels.formula.api` module to fit the model. The formula for the model will be "W ~ R + RA", which means that we are modeling wins as a function of runs scored and runs allowed. 

After fitting the model, let's print out what the coefficients are for the intercept and the slopes. Can you write the regression equation in the form of W = b0 + b1*R + b2*RA, where b0 is the intercept, b1 is the slope for runs scored, and b2 is the slope for runs allowed? 

From this model, can you answer the following questions:

1. Approximately how many additional runs do you need to score for an additional win, holding runs allowed constant?
2. Approximately how many additional runs allowed lead to an additional loss holding runs scored constant?


```python

lm_multiple = smf.ols(formula = "W ~ R + RA", data = teams_162).fit()

coefficients =  lm_multiple.params

print(coefficients)

print(1/coefficients["R"])
print(-1/coefficients["RA"])
```

**Answers**

$\hat{y}_w =  20 + 0.1 \cdot x_R - 0.1 \cdot x_{RA}$,  
where $y_w$ is wins, $x_R$ is runs scored, and $x_{RA}$ is runs allowed. 

1. For the first question, we can calculate the number of additional runs needed for an additional win by taking the reciprocal of the slope coefficient for runs scored. The slope coefficient for runs scored is 0.1, so the number of additional runs needed for an additional win is approximately 10 runs (1/0.1). 

2. For the second question, we can calculate the number of additional runs allowed that lead to an additional loss by taking the negative reciprocal of the slope coefficient for runs allowed. The slope coefficient for runs allowed is -0.1, so the number of additional runs allowed that lead to an additional loss is approximately 10 runs (-1/-0.1).



## 3.2 Finding the RMSE for the multiple regression model

Can you calculate the RMSE for this multiple regression model using the same method we used for the simple linear regression model? 

We will first use the `predict` method of the fitted model to get the predicted wins for all teams in the `teams_162` DataFrame, and then we will calculate the residuals by subtracting the predicted wins from the actual wins. Finally, we will calculate the RMSE using the formula we discussed earlier. 



```python

predictions = lm_multiple.predict(teams_162)

residuals = teams_162.W - predictions

RMSE = np.sqrt(np.mean(residuals ** 2))

RMSE


```
