## 3. Array data

Numpy is a package that provides support for working with arrays. Numpy arrays (often called "ndarray") are similar to lists, but all elements in a numpy array must be of the same type (e.g., all integers, or all floats).

Pandas Series are similar to numpy arrays, but they have an index that allows you to label the elements in the Series. All operations that work on numpy arrays also work on pandas Series. 


### 3.1 Array operations

We can easily add, subtract, and divide ndarrays of numbers that are the same length. We can also apply mathematical operations to arrays, such as `np.sum()`, `np.mean()`, `np.diff()`, etc.

**Question 3.1** Below are lists of Ohtani's seasons, earned runs, and innings pitched. Can you convert these lists to numpy arrays and calculate Ohtani's ERA (earned run average) for each season? The formula for ERA is: $$ERA = \frac{earned\_runs}{innings\_pitched} \times 9$$


```python
import numpy as np

# Shohei Ohtani pitching stats by season

# Lists showing Ohtani's seasons, earned runs, and innings pitched
seasons = [2018, 2020, 2021, 2022, 2023, 2025]
earned_runs = [19, 7, 46, 43, 46, 15]
innings_pitched = [51.667, 1.667, 130.333, 166.0, 132.0, 47.0]


# Answer: convert the lists to numpy arrays

earned_runs_array = np.array(earned_runs)
innings_pitched_array = np.array(innings_pitched)

ERA = (earned_runs_array / innings_pitched_array) * 9

np.round(ERA, 2)
```

### 4.7 Boolean indexing and the .query() methods

As we discussed extensively in YData, you can use Boolean indexing and/or the `.query()` method to extract a DataFrame that only contains rows that meet particular conditions. For example, if we have a DataFrame called `my_df` that has a column called `vals` that has quantitative data, you could extract a DataFrame that only has rows where `vals` are greater than 10 using either:

- `my_df[my_df["vals"] > 10]`

### 4.8 Adding and renaming columns

We can add new columns to a DataFrame using the syntax `my_df["new_col_name"] = values`. The values can be a list, a numpy array, or a pandas Series which must be the same length as the DataFrame. 

We can also rename columns using the `.rename()` method. For example, we can rename the column "old_col_name" to "new_col_name" using the syntax `my_df.rename(columns = {"old_col_name": "new_col_name"})`.

### 4.9 Aggregate statistics by group

We can use the pandas `groupby()` and `.agg()` method to calculate aggregate statistics separately for different groups. For example, we can had a DataFrame called `my_df` that had a categorical column called "group_col" and a quantitative column called "quant_col", we could calculate the sum of the values in the "quant_col" separately for each unique value in the "group_col" using `my_df.groupby("group_col").agg(sum_vals = ("quant_col", "sum"))`

**Question 4.9** Please calculate how many games in total each major league baseball team played. Also, sort the values such that the team that has the most values is at the top of this aggregated DataFrame. 



```python

batting.groupby("teamID").agg(total_games = ("G", "sum")).sort_values("total_games", ascending = False)
```

### 4.10 Joining/merging DataFrames

We can combine DataFrames together using the `.join()` and `.merge()` methods. The `join()` method joins to DataFrames based on the Index values, while the `.merge()` method joins two DataFrames based on values in columns in the two DataFrames. If we have two DataFrames called `df1` and `df2` that have the same Index values, we can join the using `df1.join(df2)`. Likewise, if the two DataFrame have columns with similar values (but difference Indexes) we can join them using `df1.merge(df2, left_on = "col_df1", right_on = "col_df2)`. We can also use the `how` argument to do, left, right, inner and full joins.
